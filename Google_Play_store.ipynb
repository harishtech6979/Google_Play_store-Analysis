```python
category_count= df["Category"].value_counts()
plt.figure(figsize=(14,8))
sns.barplot(
    x=category_count.head(10).values,
    y=category_count.head(10).index
    )
plt.title("Top 10 App Category")
plt.ylabel("Category")
plt.xlabel("Number of Apps")
plt.show()
```

```python
plt.figure(figsize=(14,8))
sns.histplot(df["Rating"],
             bins=20)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```

```python
avg_Rating=df.groupby("Category")["Rating"].mean()
plt.figure(figsize=(14,8))
sns.barplot(
    x=avg_Rating.head(10).values,
    y=avg_Rating.head(10).index
)
plt.title("Average Rating by Category")
plt.xlabel("Average Rating")
plt.ylabel("Category")
plt.show()
```

```python
type_count=df["Type"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(
    type_count.values,
    labels=type_count.index,
    autopct='%2.1f%%',
    colors=sns.color_palette("pastel")
)
plt.title("Distribution of App Types")
plt.show()
```

```python
content_Rating_count=df["Content Rating"].value_counts()
plt.figure(figsize=(14,8))
sns.barplot(
    x=content_Rating_count.head(10).values,
    y=content_Rating_count.head(10).index
)
plt.title("Content Rating Distribution")
plt.xlabel("No. Of Apps")
plt.ylabel("Content Rating")
plt.show()
```

```python
top_apps=df.sort_values(by="Installs",ascending=False).head(10)
plt.figure(figsize=(14,8))
sns.barplot(
    x=top_apps["Installs"],
    y=top_apps["App"]

)
plt.title("Top 10 APP by installs")
plt.xlabel("No. of Installs")
plt.ylabel("Apps")
plt.show()
```

```python
plt.figure(figsize=(14,8))
sns.scatterplot(
    x=df["Reviews"],
    y=df["Installs"]
)
plt.title("Reviews vs Installs")
plt.xlabel("Reviews")
plt.ylabel("Installs")
plt.show()
```

```python
paid_apps = df[df["Type"] == "Paid"]

plt.figure(figsize=(10, 6))

sns.histplot(
    paid_apps["Price"],
    bins=30,
    kde=True
)

plt.title("Price Distribution of Paid Apps")
plt.xlabel("Price ($)")
plt.ylabel("Number of Apps")

plt.show()
```

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'], format='mixed', dayfirst=True, errors='coerce')

yearly_apps = (
    df["Last Updated"]
    .dt.year
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12, 6))

plt.plot(
    yearly_apps.index,
    yearly_apps.values,
    marker="o"
)

plt.title("Apps Updated by Year")
plt.xlabel("Year")
plt.ylabel("Number of Apps")

plt.grid(True)

plt.show()
```

```python
# Clean and convert 'Reviews' to numeric
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

# Clean and convert 'Installs' to numeric
df['Installs'] = df['Installs'].apply(lambda x: x.replace('+', '') if '+' in str(x) else x)
df['Installs'] = df['Installs'].apply(lambda x: x.replace(',', '') if ',' in str(x) else x)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

# Clean and convert 'Price' to numeric
df['Price'] = df['Price'].apply(lambda x: x.replace('$', '') if '$' in str(x) else x)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Drop rows with NaN values introduced by coercion in these columns
df.dropna(subset=['Reviews', 'Installs', 'Price'], inplace=True)


numeric_data = df[
    ["Rating", "Reviews", "Installs", "Price"]
]

correlation = numeric_data.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Category
category_count = df["Category"].value_counts().head(10)

sns.barplot(
    x=category_count.values,
    y=category_count.index,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Top 10 App Categories")
axes[0, 0].set_xlabel("Number of Apps")


# 2. Rating
sns.histplot(
    df["Rating"],
    bins=20,
    kde=True,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Rating Distribution")
axes[0, 1].set_xlabel("Rating")


# 3. Free vs Paid
type_count = df["Type"].value_counts()

axes[1, 0].pie(
    type_count.values,
    labels=type_count.index,
    autopct="%1.1f%%"
)

axes[1, 0].set_title("Free vs Paid Apps")


# 4. Content Rating
content = df["Content Rating"].value_counts().head(5)

sns.barplot(
    x=content.values,
    y=content.index,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Content Rating")
axes[1, 1].set_xlabel("Number of Apps")


plt.tight_layout()
plt.show()
```
